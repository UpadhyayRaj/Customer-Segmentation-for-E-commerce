```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans

import warnings
warnings.filterwarnings("ignore")

plt.style.use("ggplot")
```

```python
df = pd.read_csv("../data/processed_customers.csv")

df.head()
```

```python
print(df.shape)

df.info()
```

```python
kmeans = KMeans(
    n_clusters=5,
    random_state=42
)
```

```python
kmeans.fit(df)
```

```python
clusters = kmeans.labels_

clusters
```

```python
df["Cluster"] = clusters

df.head()
```

```python
df["Cluster"].value_counts()
```

```python
centers = pd.DataFrame(
    kmeans.cluster_centers_,
    columns=df.columns[:-1]
)

centers
```

**Scatter Plot**

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="Income",
    y="Score",
    hue="Cluster",
    palette="Set1",
    s=120
)

plt.title("Customer Segments (Income vs Spending Score)")

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Score",
    hue="Cluster",
    palette="Set2",
    s=120
)

plt.title("Age vs Spending Score")

plt.show()
```

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Income",
    hue="Cluster",
    palette="Dark2",
    s=120
)

plt.title("Age vs Income")

plt.show()
```

```python
sns.pairplot(
    df,
    hue="Cluster",
    palette="tab10"
)

plt.show()
```

**Cluster-wise Average**

```python
cluster_summary = df.groupby("Cluster").mean()

cluster_summary
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    cluster_summary,
    annot=True,
    cmap="YlGnBu"
)

plt.title("Cluster Summary")

plt.show()
```

**Gender Distribution**

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Cluster",
    hue="Gender"
)

plt.title("Gender Distribution by Cluster")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    data=df,
    x="Cluster",
    y="Score"
)

plt.title("Spending Score by Cluster")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    data=df,
    x="Cluster",
    y="Income"
)

plt.title("Income Distribution by Cluster")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    data=df,
    x="Cluster",
    y="Age"
)

plt.title("Age Distribution by Cluster")

plt.show()
```

**Save Clustered Dataset**

```python
df.to_csv(
    "../data/customer_clusters.csv",
    index=False
)

print("Clustered dataset saved successfully!")
```

```python
profiles = df.groupby("Cluster").agg({

    "Age":"mean",

    "Income":"mean",

    "Score":"mean"

}).round(2)

profiles
```

```python
cluster_names = {

    0:"Premium Customers",

    1:"Budget Customers",

    2:"Average Customers",

    3:"Young High Spenders",

    4:"Regular Customers"

}

df["Segment"] = df["Cluster"].map(cluster_names)

df.head()
```

```python
df[["Cluster","Segment"]].drop_duplicates()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="Segment",
    order=df["Segment"].value_counts().index
)

plt.xticks(rotation=20)

plt.show()
```

```python
segment_summary = df.groupby("Segment").mean()

segment_summary
```

```python
df.to_csv(
    "../data/final_kmeans_output.csv",
    index=False
)

print("Final file saved successfully!")
```
